```python
import polars as pl

df = pl.read_csv("../movie_dataset/wiki_movie_plots_deduped.csv")
print(df.head())
```

```python
processed_csv = pl.read_csv("../movie_dataset/processed_movie_plots.csv")
print(processed_csv.head())
```

```python
# Number of words in each plot
df = df.with_columns(pl.col("Plot").str.split(" ").list.len().alias("Word_Count"))
print(df.select(pl.col("Word_Count")).describe())
```

```python
list_of_titles = df.select(pl.col("Title"))
for title in list_of_titles.to_series():
    print(title)
```
